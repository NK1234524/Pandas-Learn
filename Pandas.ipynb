```python
# Boolean filtering
Df = Data_Frame[10:20]
Df[Df['Age']>40][['Name','Age','Gender']]
```

```python
# Use of Boolean indexing
df_Df = Data_Frame[1:100]
df_Df[((df_Df["Age"]<60) |(df_Df["Blood Type"]=="A+")) & (df_Df["Insurance Provider"]=="Medicare")][["Age","Blood Type","Insurance Provider"]]
```

```python
# Filtering Methods
Data_Frame[Data_Frame["Insurance Provider"].isin(["Medicare","Blue Cross"])][1:100]
Data_Frame[Data_Frame["Insurance Provider"].str.contains("Blue Cross")][1:100]
# So to avoid the use of Logical notations we can also use the Filtering methods like above in Pandas
```

```python
# Cleaning Data 
# Most important concept 

# 1. isnull()
Sum_null_Value = Data_Frame.isnull().sum()
Sum_null_Value
# we got already a clean data here as no null value is present 

# 2. dropna()
# To clean the data we need to drop all the null values such that data would not contain any null value
# But this is not the right way because this will lead to the problem of all rows removal
# But to remove the value specifically of all the columns 
drop = Data_Frame.dropna(subset=["Insurance Provider"]) # It will drop all the null values present in the column mentioned
# To perforn the value filling in the data then we need to use the fillna
```
